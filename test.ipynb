```python
#pip install pymupdf4llm

from pathlib import Path
import pymupdf4llm

PDF_PATH = Path(__file__).parent / "data" / "B_menu_trattoria_nonna_rosa.pdf"
MD_PATH = Path(__file__).parent / "data" / "menu_trattoria.md"

# Estrarre il testo convertendo le tabelle in Markdown
testo_markdown = pymupdf4llm.to_markdown(str(PDF_PATH))

# Salvare il file Markdown per la verifica manuale
MD_PATH.write_text(testo_markdown, encoding="utf-8")
print(f"File Markdown generato in: {MD_PATH}")

```

```python
from pathlib import Path
import pymupdf4llm

# Se usi un Jupyter Notebook, usa la cartella corrente (cwd)
try:
    BASE_DIR = Path(__file__).parent
except NameError:
    BASE_DIR = Path.cwd()

PDF_PATH = BASE_DIR / "data" / "B_menu_trattoria_nonna_rosa.pdf"
MD_PATH = BASE_DIR / "data" / "menu_trattoria.md"

# Estrarre il testo convertendo le tabelle in Markdown
testo_markdown = pymupdf4llm.to_markdown(str(PDF_PATH))

# Salvare il file Markdown per la verifica manuale
MD_PATH.write_text(testo_markdown, encoding="utf-8")
print(f"File Markdown generato in: {MD_PATH}")
```
